# Camada Bronze — Ingestão dos Dados

Nesta etapa faremos a ingestão da base `history.csv`, preservando os dados conforme disponibilizados originalmente.

O objetivo da camada Bronze é manter uma cópia dos dados brutos, sem aplicação de tratamentos ou transformações, permitindo a rastreabilidade das etapas posteriores do projeto.

```python
import pandas as pd
from pathlib import Path
```

```python
pasta_origem = Path("dados/origem")
pasta_bronze = Path("dados/bronze")
```

## Leitura da base de origem

Carregaremos a base `history.csv` a partir da pasta de dados de origem.

Neste momento, os dados serão apenas importados para o ambiente de trabalho, sem aplicação de tratamentos ou alterações em seu conteúdo.

```python
arquivo_origem = pasta_origem / "history.csv"

df_bronze = pd.read_csv(arquivo_origem)

df_bronze.head()
```

## Inspeção da estrutura dos dados

Após a leitura da base, realizaremos uma inspeção inicial de sua estrutura.

Nesta etapa serão verificadas as dimensões do conjunto de dados, permitindo identificar a quantidade de registros e de variáveis disponíveis antes de qualquer tratamento.

```python
df_bronze.shape
```

### Dimensões da base

A base contém 5.634 registros e 23 variáveis, confirmando as dimensões esperadas para o conjunto de dados fornecido.

Com as dimensões verificadas, a próxima etapa consiste em identificar as variáveis presentes na base.

```python
df_bronze.columns
```

### Variáveis identificadas

Identificamos 23 variáveis na base, contendo informações cadastrais dos clientes, serviços contratados, características do relacionamento, dados financeiros, feedback e situação de cancelamento (Churn).

Após a identificação das variáveis, verificaremos como o Pandas interpretou os tipos de dados de cada coluna.

```python
df_bronze.dtypes
```

### Tipos de dados identificados

A inspeção mostra que a base possui variáveis numéricas e categóricas, representadas pelos tipos int64, float64 e object.

A variável "tenure", relacionada ao tempo de permanência do cliente, foi interpretada como "object". Esse comportamento será investigado posteriormente, durante a etapa de avaliação e tratamento da qualidade dos dados.

Na camada Bronze, os tipos identificados serão preservados conforme obtidos na leitura da base original.

### Informações gerais da base

A seguir, verificamos a estrutura geral do DataFrame, incluindo a quantidade de valores não nulos e os tipos de dados associados a cada variável.

```python
df_bronze.info()
```

### Valores não nulos e estrutura geral

A inspeção confirmou a presença de 5.634 registros e 23 variáveis na base.

A maior parte das colunas possui valores preenchidos em todos os registros. Entretanto, a variável "TotalCharges" apresenta 5.356 valores não nulos, indicando a existência de 278 valores ausentes.

Essas ocorrências serão preservadas na camada Bronze e analisaremos durante a etapa de tratamento dos dados.

## Armazenamento na camada Bronze

Após a leitura e a inspeção inicial da estrutura, os dados serão armazenados na camada Bronze.

Nesta etapa, nenhuma correção ou transformação será aplicada. O conjunto será preservado conforme obtido da fonte de origem, mantendo inclusive possíveis inconsistências que serão avaliadas posteriormente na camada Silver.

```python
arquivo_bronze = pasta_bronze / "history_bronze.csv"

df_bronze.to_csv(arquivo_bronze, index=False)
```

### Validação do armazenamento

Após a gravação, verificamos a existência do arquivo gerado na camada Bronze, confirmando a conclusão do processo de ingestão.

```python
arquivo_bronze.exists()
```

A base original foi carregada e armazenada na camada Bronze sem aplicação de tratamentos ou transformações, preservando as características dos dados recebidos.
